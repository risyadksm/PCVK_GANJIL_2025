## **SOAL NO 1**




```python
img_path = '/content/drive/MyDrive/PCVK/markzuckerberg.jpg'

img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(5,5))
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title('Citra Asli (0–255)')
plt.axis('off')
plt.show()
```

```python
# Menyempitkan rentang intensitas dari 0–255 ke 80–150
img_sempit = cv2.normalize(img, None, 80, 150, cv2.NORM_MINMAX)

plt.figure(figsize=(5,5))
plt.imshow(img_sempit, cmap='gray', vmin=0, vmax=255)
plt.title('Citra dengan Rentang Intensitas Sempit (80–150)')
plt.axis('off')
plt.show()
```

```python
img_dark_path = "/content/drive/MyDrive/PCVK/human2.jpg"

img_dark = cv2.imread(img_dark_path, cv2.IMREAD_GRAYSCALE)
img_dark = cv2.resize(img_dark, (256, 256)) # resize
img_dark_norm = cv2.normalize(img_dark, None, 0, 255, cv2.NORM_MINMAX) # normalisasi
img_dark_eq = cv2.equalizeHist(img_dark) # histogram equalization

# ==== Perbandingan Statistik ====
def get_stats(img):
    return np.mean(img), np.std(img)

mean_orig, std_orig = get_stats(img_dark)
mean_norm, std_norm = get_stats(img_dark_norm)
mean_eq, std_eq = get_stats(img_dark_eq)

print("Statistik Citra:")
print(f"- Citra Asli (Gelap)      : mean={mean_orig:.2f}, std={std_orig:.2f}")
print(f"- Setelah Normalisasi     : mean={mean_norm:.2f}, std={std_norm:.2f}")
print(f"- Setelah Equalization    : mean={mean_eq:.2f}, std={std_eq:.2f}")

# ==== Visualisasi hasil ====
plt.figure(figsize=(14,10))

# Citra
plt.subplot(3,3,1)
plt.imshow(img_dark, cmap='gray')
plt.title('Citra Asli (Dominan Gelap)')
plt.axis('off')

plt.subplot(3,3,2)
plt.imshow(img_dark_norm, cmap='gray')
plt.title('Hasil Normalisasi Histogram')
plt.axis('off')

plt.subplot(3,3,3)
plt.imshow(img_dark_eq, cmap='gray')
plt.title('Hasil Histogram Equalization')
plt.axis('off')

# Histogram
plt.subplot(3,3,4)
plt.hist(img_dark.ravel(), 256, [0,256])
plt.title('Histogram Citra Asli')

plt.subplot(3,3,5)
plt.hist(img_dark_norm.ravel(), 256, [0,256])
plt.title('Histogram Normalisasi')

plt.subplot(3,3,6)
plt.hist(img_dark_eq.ravel(), 256, [0,256])
plt.title('Histogram Equalization')

plt.tight_layout()
plt.show()
```

## **SOAL NO 3**

```python
img_3 = cv2.imread('/content/drive/MyDrive/PCVK/markzuckerberg.jpg', cv2.IMREAD_GRAYSCALE)

print("Ukuran citra (tinggi, lebar):", img_3.shape)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(img_3, cmap='gray')
plt.title("Citra Asli (Grayscale 8-bit)")
plt.axis('on')
```

```python
# --- Konversi Biner Sederhana (Global Threshold) ---
_, binary_simple = cv2.threshold(img_3, 128, 255, cv2.THRESH_BINARY)

# --- Ordered Dithering (2x2 Bayer Matrix Sederhana) ---
T = np.array([[0, 128],
              [192, 64]], dtype=np.uint8)

threshold_map = np.tile(T, (img_3.shape[0]//2 + 1, img_3.shape[1]//2 + 1))
threshold_map = threshold_map[:img_3.shape[0], :img_3.shape[1]]
ordered_dither = np.where(img_3 > threshold_map, 255, 0).astype(np.uint8)

# --- Floyd–Steinberg Error Diffusion ---
img_fs = img_3.astype(np.float32).copy()
for y in range(img_fs.shape[0]-1):
    for x in range(1, img_fs.shape[1]-1):
        old_pixel = img_fs[y, x]
        new_pixel = 255 if old_pixel > 128 else 0
        img_fs[y, x] = new_pixel
        error = old_pixel - new_pixel
        img_fs[y, x+1] += error * 7/16
        img_fs[y+1, x-1] += error * 3/16
        img_fs[y+1, x] += error * 5/16
        img_fs[y+1, x+1] += error * 1/16

img_fs = np.clip(img_fs, 0, 255).astype(np.uint8)
```

```python
# --- Crop area wajah ---
x, y, w, h = 450, 80, 300, 300
crop_gray = img_3[y:y+h, x:x+w]
crop_bin = binary_simple[y:y+h, x:x+w]
crop_ord = ordered_dither[y:y+h, x:x+w]
crop_fs  = img_fs[y:y+h, x:x+w]

# --- Visualisasi hasil ---
plt.figure(figsize=(18, 8))

titles = [
    "Citra Asli (Grayscale 8-bit)",
    "Threshold Biner (128)",
    "Ordered Dithering (2x2)",
    "Floyd–Steinberg Error Diffusion"
]
images = [img_3, binary_simple, ordered_dither, img_fs]
crops = [crop_gray, crop_bin, crop_ord, crop_fs]

for i in range(4):
    # Gambar utama
    plt.subplot(2, 4, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i], fontsize=11)
    plt.axis('on')

    # Tambahkan kotak merah penanda crop area
    plt.gca().add_patch(plt.Rectangle(
        (x, y), w, h, edgecolor='red', facecolor='none', linewidth=2
    ))

    # Gambar crop wajah di bawah
    plt.subplot(2, 4, i+5)
    plt.imshow(crops[i], cmap='gray')
    plt.title(f'Crop {titles[i]}', fontsize=10)
    plt.axis('off')

plt.tight_layout(pad=2.0, w_pad=1.0)
plt.show()
```
